```python
date_columns = [
    'PQ First Sent to Client Date',
    'PO Sent to Vendor Date',
    'Scheduled Delivery Date',
    'Delivered to Client Date',
    'Delivery Recorded Date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce', dayfirst=True)

print("Date conversion completed!")

print(df[date_columns].dtypes)
```

```python
df['Delivery Delay (Days)'] = (
    df['Delivered to Client Date'] -
    df['Scheduled Delivery Date']
).dt.days

print(df[['Scheduled Delivery Date',
          'Delivered to Client Date',
          'Delivery Delay (Days)']].head(10))
```

```python
df['Delivery Status'] = df['Delivery Delay (Days)'].apply(
    lambda x: 'Missing Date' if pd.isna(x)
    else 'Delayed' if x > 0
    else 'On Time' if x == 0
    else 'Early'
)

print(df['Delivery Status'].value_counts())
```

```python
total_shipments = len(df)

delivered = df['Delivery Status'].isin(
    ['Delayed', 'On Time', 'Early']
).sum()

delayed = (df['Delivery Status'] == 'Delayed').sum()
on_time = (df['Delivery Status'] == 'On Time').sum()
early = (df['Delivery Status'] == 'Early').sum()

delivery_rate = (delivered / total_shipments) * 100
delay_rate = (delayed / delivered) * 100 if delivered else 0
on_time_rate = (on_time / delivered) * 100 if delivered else 0

print("Total Shipments:", total_shipments)
print("Delivered Shipments:", delivered)
print("Delayed Shipments:", delayed)
print("On-Time Shipments:", on_time)
print("Early Shipments:", early)
print("Delivery Rate:", round(delivery_rate, 2), "%")
print("Delay Rate:", round(delay_rate, 2), "%")
print("On-Time Rate:", round(on_time_rate, 2), "%")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("SCMS_Delivery_History_Dataset.csv")
```

```python
country_analysis = df['Country'].value_counts().head(10)

plt.figure(figsize=(10, 6))

plt.barh(country_analysis.index, country_analysis.values)

plt.title('Top 10 Countries by Shipment Volume')
plt.xlabel('Number of Shipments')
plt.ylabel('Country')

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
print(df.columns.tolist())
```

```python
df['Scheduled Delivery Date'] = pd.to_datetime(
    df['Scheduled Delivery Date'], errors='coerce'
)

df['Delivered to Client Date'] = pd.to_datetime(
    df['Delivered to Client Date'], errors='coerce'
)

df['Delivery Status'] = 'On Time'

df.loc[
    df['Delivered to Client Date'] < df['Scheduled Delivery Date'],
    'Delivery Status'
] = 'Early'

df.loc[
    df['Delivered to Client Date'] > df['Scheduled Delivery Date'],
    'Delivery Status'
] = 'Delayed'

print(df['Delivery Status'].value_counts())
```

```python
delivery_mode = pd.crosstab(
    df['Shipment Mode'],
    df['Delivery Status']
)

delivery_mode.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Delivery Performance by Shipment Mode')
plt.xlabel('Shipment Mode')
plt.ylabel('Number of Shipments')
plt.xticks(rotation=45)
plt.legend(title='Delivery Status')
plt.tight_layout()
plt.show()
```

```python
delivery_percentage = pd.crosstab(
    df['Shipment Mode'],
    df['Delivery Status'],
    normalize='index'
) * 100

print(delivery_percentage.round(2))
```

```python
df['Freight Cost (USD)'] = pd.to_numeric(
    df['Freight Cost (USD)'],
    errors='coerce'
)

print(df['Freight Cost (USD)'].dtype)
```

```python
df['Delay Days'] = (
    df['Delivered to Client Date'] -
    df['Scheduled Delivery Date']
).dt.days

print(df[['Vendor', 'Delay Days']].head(10))
```

```python
vendor_delay = (
    df.groupby('Vendor')['Delay Days']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

plt.barh(vendor_delay.index, vendor_delay.values)

plt.title('Top 10 Vendors by Average Delivery Delay')
plt.xlabel('Average Delay (Days)')
plt.ylabel('Vendor')

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
vendor_performance = df.groupby('Vendor').agg(
    Total_Shipments=('Vendor', 'count'),
    Average_Delay=('Delay Days', 'mean'),
    Delayed_Shipments=('Delivery Status',
                       lambda x: (x == 'Delayed').sum())
)

vendor_performance['Delay_Rate'] = (
    vendor_performance['Delayed_Shipments'] /
    vendor_performance['Total_Shipments'] * 100
)

vendor_performance = vendor_performance.sort_values(
    by='Delay_Rate', ascending=False
)

display(vendor_performance.head(10).round(2))
```

```python
country_performance = df.groupby('Country').agg(
    Total_Shipments=('Country', 'count'),
    Delayed_Shipments=('Delivery Status',
                       lambda x: (x == 'Delayed').sum())
)

country_performance['Delay_Rate'] = (
    country_performance['Delayed_Shipments'] /
    country_performance['Total_Shipments'] * 100
)

country_performance = country_performance.sort_values(
    by='Delay_Rate', ascending=False
)

display(country_performance.head(10).round(2))
```

```python
freight_mode = df.groupby('Shipment Mode')['Freight Cost (USD)'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,6))

plt.bar(
    freight_mode.index,
    freight_mode.values
)

plt.title('Average Freight Cost by Shipment Mode')
plt.xlabel('Shipment Mode')
plt.ylabel('Average Freight Cost (USD)')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
total_shipments = len(df)

delayed = (df['Delivery Status'] == 'Delayed').sum()
on_time = (df['Delivery Status'] == 'On Time').sum()
early = (df['Delivery Status'] == 'Early').sum()

print("Total Shipments:", total_shipments)
print("Delayed Shipments:", delayed)
print("On-Time Shipments:", on_time)
print("Early Shipments:", early)

print("Delay Percentage:",
      round(delayed / total_shipments * 100, 2), "%")

print("On-Time Percentage:",
      round(on_time / total_shipments * 100, 2), "%")
```

# Business Insights and Recommendations

## Key Business Insights

1. **Delivery Performance:** Analyse the proportion of early, on-time and delayed shipments to evaluate delivery reliability.

2. **Geographical Distribution:** Identify countries with high shipment volumes and destinations with relatively high delay rates.

3. **Vendor Performance:** Compare vendor shipment volumes, average delivery delays and delay percentages to identify areas requiring operational improvement.

4. **Transportation Efficiency:** Compare delivery performance and freight expenditure across different shipment modes.

5. **Freight Cost Analysis:** Examine freight expenditure across transportation modes and delivery categories to identify potential cost-optimization opportunities.

## Business Recommendations

- Investigate recurring delivery delays and identify their operational causes.
- Review vendor performance regularly using shipment volume and delay-rate indicators.
- Evaluate transportation methods based on both cost and delivery reliability.
- Prioritize logistics planning for countries with high shipment volumes.
- Monitor freight expenditure to identify potential cost-saving opportunities.
